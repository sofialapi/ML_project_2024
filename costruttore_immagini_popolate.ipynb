```python
import db_interactions as dbi
import image_manager as im
import db_manager as dbm
import pandas as pd
import csv
```

```python
database_crateri= pd.read_csv('crateri_marte_filtered.csv')
```

```python
'''
matrice originale: 8192x32768
singola matrice: 512x512
latitudine: +45 -> -45
longitudine: -180 -> +180
incremento latitudine: -5.625
incremento longitudine: +5.625
incremento centro matrice x&y: +512
'''
lat_iniziale = 45
lon_iniziale = -180
incremento_latitudine = -5.625
incremento_longitudine = +5.625
centro_iniziale_x = 256
centro_iniziale_y = 256
incremento_centro = 512

for i in range(16):
    #stabilisco il centroide su y dell'immagine e le latitudini max e min relative
    centro_matrice_y= centro_iniziale_y + i*incremento_centro
    latitudine_minima = lat_iniziale + i*incremento_latitudine
    latitudine_massima = lat_iniziale + (i+1)*incremento_latitudine
    for j in range(64):
       #stabilisco il centroide su x dell'immagine e le longitudini max e min relative
       centro_matrice_x = centro_iniziale_x + j*incremento_centro
       longitudine_minima = lon_iniziale + j*incremento_longitudine
       longitudine_massima = lon_iniziale + (j+1)*incremento_longitudine
       #estraggo e salvo l'immagine 512x512 selezionata
       #da qualche parte c'è un inversione degli indici x,y non è un problema perchè siamo persone adulte ed equilibrate. 
       dati_immagine=dbi.estrai_sotto_matrice(centro_matrice_y, centro_matrice_x, 512)
       immagine=im.salva_sotto_matrice_come_png(dati_immagine, f"immagine_{i}_{j}.png", 'immagini_sam')
       #filtro il database iniziale ci crateri e ottengo solo i crateri che si trovano in quell'immagine
       crateri_immagine=dbm.filter_lon(database_crateri,longitudine_minima, longitudine_massima)
       crateri_immagine=dbm.filter_lat(crateri_immagine, latitudine_massima, latitudine_minima)
       crateri_immagine.to_csv(f'csv_sam/crateri_{i}_{j}.csv')
    print(f"{i}_{j}")
```
