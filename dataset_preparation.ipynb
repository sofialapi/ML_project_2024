```python
import pandas as pd
import numpy as np
import shutil
```

creazione dei file txt delle bounding box con il nome uguale all'immagine


```python
def convert_csv_to_txt(csv_file, i, j):
    # Leggi il file CSV
    df = pd.read_csv(csv_file)
    img_w = 512
    img_h = 512
    
    # Determina il nome del file di output
    label_file = f"txt_sam/immagine_{i}_{j}.txt"
    
    # Controlla se il DataFrame è vuoto
    if df.empty:
        # Crea un file txt vuoto
        open(label_file, "w").close()
    else:
        # Apri il file di output una volta
        with open(label_file, "w") as f:
            for index, row in df.iterrows():
                bbox_x = row['pixel_x']
                bbox_y = row['pixel_y']
                bbox_w = row['lato_bb']
                bbox_h = row['lato_bb']  # Assuming square bounding boxes
                
                # Calcola i valori normalizzati
                norm_x = bbox_x / img_w
                norm_y = bbox_y / img_h
                norm_w = bbox_w / img_w
                norm_h = bbox_h / img_h

                # Scrivi la riga nel file di output
                f.write(f"0 {norm_x} {norm_y} {norm_w} {norm_h}\n")
```

```python
for i in range(16):
    for j in range(64):
        convert_csv_to_txt(f"csv_sam/crateri_{i}_{j}.csv", i, j)
```

divisione tra immagini e file di training e di valutazione

```python
for i in range(16):
    for j in range(64):
        directory_immagine= f"immagini_sam/immagine_{i}_{j}.png"
        directory_txt= f"txt_sam/immagine_{i}_{j}.txt"
        if np.random.random() < 0.8:
            shutil.copy(directory_immagine, f"dataset/images/train/immagine_{i}_{j}.png")
            shutil.copy(directory_txt, f"dataset/labels/train/immagine_{i}_{j}.txt")
        else:
            shutil.copy(directory_immagine, f"dataset/images/val/immagine_{i}_{j}.png")
            shutil.copy(directory_txt, f"dataset/labels/val/immagine_{i}_{j}.txt")
            
            
```
